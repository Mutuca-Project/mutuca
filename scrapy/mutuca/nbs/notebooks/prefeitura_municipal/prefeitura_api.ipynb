## Utilização da API fornecida para obtenção de dados de licitações e contratos

```python
import requests
import json

url = "https://licitacoes.caruaru.pe.gov.br:8082/licitacao/exportar/json"
response = requests.get(url)


if response.status_code == 200:
    data = response.json()
    output_file = "licitacoes.json"

    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=4)
    
    print(f"Arquivo salvo como {output_file}")
else:
    print("Erro na requisição. Status Code: {response.status_code}")
```
