# Unidades Gestoras

Será útil termos à mão os códigos referentes a cada unidade gestora. A página da api trás uma tabela identificando cada unidade por seu código. Para obter essas informações eu busquei o arquivo que retorna estas informações e fiz uma cópia simples do arquivo para o ambiente local. 

![xpath](../assets/cityhall_assets/cod_unidade_gestora_response.png)

## Formatação do dicionário

```python
import os
import requests
import json
import csv
```

```python
base_path = os.getcwd()
file = 'dicionario.json'
dicionario_path = os.path.join(base_path, file)

with open(dicionario_path, 'r') as f:
    dicionario = json.load(f)

dicionario.keys()
```

```python
json_data = dicionario.get('pojo')

unidades = json_data.get('unidades')

unidades_list = []
for unidade in unidades:
    unidades_dicionario = {
        unidade['codigo']: unidade['denominacaoPara'].replace('para', '').strip(),
    }
    unidades_list.append(unidades_dicionario)


unidades_list
```

## Salvando o dicionário em um CSV local

```python
with open('codigo_unidades.csv', mode='w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)

    writer.writerow(['cod_unidade_gestora', 'nome_unidade_gestora'])

    for unidade_dict in unidades_list:
        for cod_unidade, nome_unidade in unidade_dict.items():
            writer.writerow([cod_unidade, nome_unidade])
    print("Dicionário de Unidades Gestoras exportadas com sucesso para 'codigo_unidades.csv'")
```
